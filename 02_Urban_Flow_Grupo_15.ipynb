En este punto se deben mostrar las imágenes originales. Cada imágen debe tener un ancho de 10 pulgas y un alto de 8 pulgadas, se debe mostrar el nombre de la imagen.



````python
#@title Listado de imágenes disponibles

ruta_imagenes: Path = Path("urban_flow/data/raw/imgs")

extensiones_validas: set[str] = {
  ".jpg", ".jpeg", ".png", ".jp2", ".qoi"
}

archivos_imagenes: list[Path] = [
  archivo for archivo in ruta_imagenes.iterdir()
  if archivo.is_file() and archivo.suffix.lower() in extensiones_validas
]

datos_imagenes: list[dict[str, Any]] = []

for archivo in archivos_imagenes:
  datos_imagenes.append({
    "filename": archivo.name,
    "size_kb": round(archivo.stat().st_size / 1024, 2)
  })

df_imagenes: pd.DataFrame = pd.DataFrame(datos_imagenes)
df_imagenes
````

````python
#@title Separación de imágenes en plates y completes

group_images: dict[str, list[dict[str, Any]]] = {
  "plates": [],
  "completes": []
}

for archivo in archivos_imagenes:
  with Image.open(archivo) as imagen:
    ancho, alto = imagen.size

  grupo: str = "plates" if "patente" in archivo.stem else "completes"

  group_images[grupo].append({
    "filename": archivo.name,
    "width": ancho,
    "height": alto,
    "area": ancho * alto,
    "path": str(archivo),
    "patent": None
  })

print(f"El grupo 'plates' contiene {len(group_images['plates'])} imágenes.")
print(f"El grupo 'completes' contiene {len(group_images['completes'])} imágenes.")
````

````python
#@title Resolución promedio por grupo

datos_resolucion: list[dict[str, Any]] = []

for grupo, imagenes in group_images.items():
  df_grupo: pd.DataFrame = pd.DataFrame(imagenes)

  datos_resolucion.append({
    "grupo": grupo,
    "cantidad": len(df_grupo),
    "ancho_promedio": round(df_grupo["width"].mean(), 2),
    "alto_promedio": round(df_grupo["height"].mean(), 2),
    "area_promedio": round(df_grupo["area"].mean(), 2)
  })

df_resolucion: pd.DataFrame = pd.DataFrame(datos_resolucion)
df_resolucion
````

````python
#@title Guardar diccionario group_images en JSON

ruta_json: Path = Path("ubran_flow/data/interim/group_images.json")
ruta_json.parent.mkdir(parents=True, exist_ok=True)

with open(ruta_json, "w", encoding="utf-8") as archivo_json:
  json.dump(group_images, archivo_json, indent=2, ensure_ascii=False)

print(f"Diccionario group_images guardado en {ruta_json}")
````

````python
#@title Función para mostrar imágenes aleatorias

def mostrar_imagenes_aleatorias(
  imagenes_agrupadas: dict[str, list[dict[str, Any]]],
  cantidad_por_grupo: int = 4
) -> None:
  """
  Muestra imágenes originales aleatorias en una tabla de 2 columnas y 4 filas.
  """

  imagenes_plates: list[dict[str, Any]] = random.sample(
    imagenes_agrupadas["plates"],
    cantidad_por_grupo
  )

  imagenes_completes: list[dict[str, Any]] = random.sample(
    imagenes_agrupadas["completes"],
    cantidad_por_grupo
  )

  imagenes_muestra: list[dict[str, Any]] = []

  for indice in range(cantidad_por_grupo):
    imagenes_muestra.append(imagenes_plates[indice])
    imagenes_muestra.append(imagenes_completes[indice])

  figura, ejes = plt.subplots(4, 2, figsize=(20, 32))
  ejes_planos = ejes.flatten()

  for eje, datos_imagen in zip(ejes_planos, imagenes_muestra):
    imagen = Image.open(datos_imagen["path"])
    eje.imshow(imagen)
    eje.set_title(datos_imagen["filename"])
    eje.axis("off")

  plt.tight_layout()
  plt.show()
````

En este punto realizaremos un procesamiento de las imágenes, en cada punto se debe llamar a la función que muestra las imágenes trabajadas en el punto anterior.

- Conversión a escala de grises de las imágenes originales y almacenarlas en `urban_flow/data/interim/imgs/03_01_gray_scale/plates` y `urban_flow/data/interim/imgs/03_01_gray_scale/completes`, según la clasificación del ejercicio 2.

````python
#@title Conversión a escala de grises

def procesar_grises(
  imagenes: list[dict[str, Any]],
  ruta_salida: Path
) -> list[dict[str, Any]]:
  """
  Convierte imágenes a escala de grises y las guarda.

  Args:
    imagenes: Lista de dicts con metadata de imagen.
    ruta_salida: Directorio donde guardar las imágenes.

  Returns:
    Lista de dicts actualizados con la nueva ruta.
  """
  ruta_salida.mkdir(parents=True, exist_ok=True)
  imagenes_grises: list[dict[str, Any]] = []

  for datos in imagenes:
    ruta_completa = Path(
      "/content/lcd-tp-urban-flow"
    ) / datos["path"]

    img = cv2.imread(str(ruta_completa))

    if img is None:
      img_pil = Image.open(ruta_completa).convert("RGB")
      img = np.array(img_pil)
      img = cv2.cvtColor(img, cv2.COLOR_RGB2BGR)

    gris = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    nombre_jpg = Path(datos["filename"]).stem + ".jpg"
    ruta_destino = ruta_salida / nombre_jpg
    cv2.imwrite(str(ruta_destino), gris)

    imagenes_grises.append({
      **datos,
      "filename": nombre_jpg,
      "path": str(ruta_destino)
    })

  return imagenes_grises


def mostrar_imagenes_aleatorias(
  imagenes_agrupadas: dict[str, list[dict[str, Any]]],
  cantidad_por_grupo: int = 4
) -> None:
  """
  Muestra imágenes aleatorias en una tabla de 2 columnas y 4 filas.

  Args:
    imagenes_agrupadas: Diccionario con grupos de imágenes.
    cantidad_por_grupo: Cantidad de imágenes por grupo a mostrar.
  """
  imagenes_plates: list[dict[str, Any]] = random.sample(
    imagenes_agrupadas["plates"],
    cantidad_por_grupo
  )
  imagenes_completes: list[dict[str, Any]] = random.sample(
    imagenes_agrupadas["completes"],
    cantidad_por_grupo
  )

  imagenes_muestra: list[dict[str, Any]] = []
  for indice in range(cantidad_por_grupo):
    imagenes_muestra.append(imagenes_plates[indice])
    imagenes_muestra.append(imagenes_completes[indice])

  figura, ejes = plt.subplots(4, 2, figsize=(20, 32))
  ejes_planos = ejes.flatten()

  for eje, datos_imagen in zip(ejes_planos, imagenes_muestra):
    imagen = Image.open(datos_imagen["path"])
    img_array = np.array(imagen)
    es_gris = len(img_array.shape) == 2
    eje.imshow(img_array, cmap="gray" if es_gris else None)
    eje.set_title(datos_imagen["filename"])
    eje.axis("off")

  plt.tight_layout()
  plt.show()


gray_plates = procesar_grises(
  group_images["plates"],
  Path(
    "/content/lcd-tp-urban-flow/urban_flow/data"
    "/interim/imgs/03_01_gray_scale/plates"
  )
)
gray_completes = procesar_grises(
  group_images["completes"],
  Path(
    "/content/lcd-tp-urban-flow/urban_flow/data"
    "/interim/imgs/03_01_gray_scale/completes"
  )
)

mostrar_imagenes_aleatorias(
  {"plates": gray_plates, "completes": gray_completes}
)
````

````python
#@title Suavizado (blur)

def aplicar_blur(
  imagenes: list[dict[str, Any]],
  ruta_salida: Path,
  kernel: int = 5
) -> list[dict[str, Any]]:
  """
  Aplica suavizado gaussiano a imágenes en escala de grises.

  Args:
    imagenes: Lista de dicts con metadata de imagen.
    ruta_salida: Directorio donde guardar las imágenes suavizadas.
    kernel: Tamaño del kernel gaussiano (debe ser impar).

  Returns:
    Lista de dicts actualizados con la nueva ruta.
  """
  ruta_salida.mkdir(parents=True, exist_ok=True)
  imagenes_blur: list[dict[str, Any]] = []

  for datos in imagenes:
    img = cv2.imread(datos["path"], cv2.IMREAD_GRAYSCALE)
    blur = cv2.GaussianBlur(img, (kernel, kernel), 0)
    ruta_destino = ruta_salida / datos["filename"]
    cv2.imwrite(str(ruta_destino), blur)
    imagenes_blur.append({**datos, "path": str(ruta_destino)})

  return imagenes_blur


blur_plates = aplicar_blur(
  gray_plates,
  Path(
    "/content/lcd-tp-urban-flow/urban_flow/data"
    "/interim/imgs/03_02_blur/plates"
  ),
  kernel=11
)
blur_completes = aplicar_blur(
  gray_completes,
  Path(
    "/content/lcd-tp-urban-flow/urban_flow/data"
    "/interim/imgs/03_02_blur/completes"
  ),
  kernel=15
)

mostrar_imagenes_aleatorias(
  {"plates": blur_plates, "completes": blur_completes}
)
````

````python
#@title Detección de bordes (Canny)

def aplicar_canny(
  imagenes: list[dict[str, Any]],
  ruta_salida: Path,
  umbral1: int = 50,
  umbral2: int = 150
) -> list[dict[str, Any]]:
  """
  Aplica detección de bordes Canny a imágenes suavizadas.

  Args:
    imagenes: Lista de dicts con metadata de imagen.
    ruta_salida: Directorio donde guardar las imágenes procesadas.
    umbral1: Umbral inferior para la histéresis.
    umbral2: Umbral superior para la histéresis.

  Returns:
    Lista de dicts actualizados con la nueva ruta.
  """
  ruta_salida.mkdir(parents=True, exist_ok=True)
  imagenes_canny: list[dict[str, Any]] = []

  for datos in imagenes:
    img = cv2.imread(datos["path"], cv2.IMREAD_GRAYSCALE)
    bordes = cv2.Canny(img, umbral1, umbral2)
    ruta_destino = ruta_salida / datos["filename"]
    cv2.imwrite(str(ruta_destino), bordes)
    imagenes_canny.append({**datos, "path": str(ruta_destino)})

  return imagenes_canny


canny_plates = aplicar_canny(
  blur_plates,
  Path(
    "/content/lcd-tp-urban-flow/urban_flow/data"
    "/interim/imgs/03_03_canny/plates"
  )
)
canny_completes = aplicar_canny(
  blur_completes,
  Path(
    "/content/lcd-tp-urban-flow/urban_flow/data"
    "/interim/imgs/03_03_canny/completes"
  )
)

mostrar_imagenes_aleatorias(
  {"plates": canny_plates, "completes": canny_completes}
)
````

````python
#@title Definición de la función extraer_patente

import easyocr
def extraer_patente(image_path) -> str:
  """ Extrae la patente de una imagen.

  Args:
      image_path (str): Ruta de la imagen.

  Returns:
      str: Patente extraída.
  """
  reader = easyocr.Reader(['es'])
  detection:list = reader.readtext(image_path)
  if len(detection) == 0:
    return ''
  patent = ''
  for d in detection:
    if isinstance(d,tuple) and len(d) == 3:
      if isinstance(d[1],str):
        patent+=d[1]
  return patent
````

````python
#@title Normalización de texto OCR

def normalizar_patente_ocr(texto_patente: str) -> str:
  """
  Normaliza el texto obtenido por OCR.
  """
  texto_limpio: str = re.sub(
    r"[^A-Z0-9]",
    "",
    texto_patente.upper()
  )

  return texto_limpio
````

````python
#@title Extracción de patentes desde imágenes plates

ruta_temporal_ocr: Path = Path(
  "urban_flow/data/interim/imgs/04_ocr_temp"
)
ruta_temporal_ocr.mkdir(parents=True, exist_ok=True)

for datos_imagen in group_images["plates"]:
  ruta_imagen: Path = Path(datos_imagen["path"])

  if not ruta_imagen.exists():
    datos_imagen["patent"] = ""
    continue

  ruta_ocr: Path = ruta_imagen

  if ruta_imagen.suffix.lower() not in [".jpg", ".jpeg", ".png"]:
    try:
      imagen_pil: Image.Image = Image.open(ruta_imagen).convert("RGB")
      ruta_ocr = ruta_temporal_ocr / f"{ruta_imagen.stem}.jpg"
      imagen_pil.save(ruta_ocr)
    except Exception:
      datos_imagen["patent"] = ""
      continue

  try:
    patente_detectada: str = extraer_patente(str(ruta_ocr))
    datos_imagen["patent"] = normalizar_patente_ocr(patente_detectada)
  except Exception:
    datos_imagen["patent"] = ""

print("Patentes extraídas y almacenadas en group_images['plates'].")
````

````python
#@title Función para calcular coincidencia entre patentes

def calcular_ratio_patente(
  patente_original: str,
  patente_detectada: str
) -> float:
  """
  Calcula el porcentaje de coincidencia entre dos patentes.
  """
  ratio: float = SequenceMatcher(
    None,
    patente_original,
    patente_detectada
  ).ratio()

  return round(ratio * 100, 2)
````

````python
#@title Relación entre patentes OCR y dataset de multas

coincidencias: list[dict[str, Any]] = []

for _, multa in df_multas.iterrows():
  patente_original: str = str(multa["patente"])

  mejor_ratio: float = 0.0
  mejor_patente_imagen: str = ""
  mejor_imagen: str = ""

  for imagen in group_images["plates"]:
    patente_imagen: str = str(imagen["patent"])

    if patente_imagen == "":
      continue

    ratio: float = calcular_ratio_patente(
      patente_original,
      patente_imagen
    )

    if ratio > mejor_ratio:
      mejor_ratio = ratio
      mejor_patente_imagen = patente_imagen
      mejor_imagen = imagen["path"]

  if mejor_ratio >= 80:
    registro: dict[str, Any] = multa.to_dict()
    registro["imagen"] = mejor_imagen
    registro["patente_imagen"] = mejor_patente_imagen
    registro["ratio"] = mejor_ratio
    coincidencias.append(registro)

df_multas_imagen: pd.DataFrame = pd.DataFrame(coincidencias)

df_multas_imagen.head()
````
