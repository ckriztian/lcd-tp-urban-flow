## Ejercicio 03 - Diseño del modelo lógico

A partir del dataset `speeding_fines_image.csv`, se identifican cuatro entidades principales: `Vehiculo`, `Multa`, `Radar` y `Evidencia`.

El modelo lógico busca representar la relación entre las infracciones registradas, los vehículos involucrados, los radares que detectaron las multas y la evidencia visual asociada.

### Entidad: Vehiculo

Representa al vehículo identificado por su patente.

| Atributo | Descripción |
|---|---|
| `patente` | Patente del vehículo |

### Entidad: Radar

Representa el dispositivo que registra la infracción.

| Atributo | Descripción |
|---|---|
| `radar_id` | Identificador del radar |
| `ubicacion` | Ubicación donde se registró la infracción |

### Entidad: Evidencia

Representa la evidencia visual asociada opcionalmente a una multa.

| Atributo | Descripción |
|---|---|
| `archivo_imagen` | Ruta o nombre del archivo de imagen |
| `patente_imagen` | Patente detectada mediante OCR |
| `ratio` | Porcentaje de coincidencia entre la patente original y la detectada |

### Entidad: Multa

Representa la infracción registrada. Es la entidad central del modelo, ya que contiene los datos de la multa y se relaciona con las demás entidades.

| Atributo | Descripción |
|---|---|
| `multa_id` | Identificador de la multa |
| `fecha` | Fecha de la infracción |
| `hora` | Hora de la infracción |
| `velocidad_registrada` | Velocidad detectada |
| `velocidad_maxima` | Velocidad máxima permitida |
| `estado_multa` | Estado de la multa |
| `exceso_velocidad_real` | Exceso real de velocidad |
| `exceso_velocidad` | Exceso de velocidad calculado |
| `vehiculo` | Vehículo asociado a la multa |
| `radar` | Radar que registró la multa |
| `evidencia` | Evidencia visual asociada, si existe |

### Relaciones consideradas

| Relación | Cardinalidad | Descripción |
|---|---|---|
| Vehiculo - Multa | 1 a N | Un vehículo puede tener muchas multas |
| Radar - Multa | 1 a N | Un radar puede generar muchas multas |
| Multa - Evidencia | 1 a 0..1 | Una multa puede tener una evidencia opcional |

### Representación lógica

```text
Vehiculo (patente) 1 ──── N Multa N ──── 1 Radar (radar_id)
                              |
                              |
                           0..1
                              |
                          Evidencia

El modelo lógico se diseñó tomando como referencia las columnas disponibles en `speeding_fines_image.csv`. La entidad `Vehiculo` se identifica mediante la patente, ya que el archivo no contiene un identificador propio para vehículo. La entidad `Multa` contiene los datos principales de la infracción, se relaciona con `Vehiculo` mediante la patente y con `Radar` mediante `radar_id`. La entidad `Evidencia` representa la información visual asociada a una multa y su relación es opcional.

````python
#@title Modelo lógico del sistema de multas de tránsito

class Vehiculo:
  """
  Representa un vehículo involucrado en una infracción.
  """

  def __init__(self, patente: str) -> None:
    self.patente: str = patente


class Radar:
  """
  Representa un radar que detecta infracciones.
  """

  def __init__(self, radar_id: str, ubicacion: str) -> None:
    self.radar_id: str = radar_id
    self.ubicacion: str = ubicacion


class Evidencia:
  """
  Representa la evidencia visual asociada a una multa.
  """

  def __init__(
    self,
    archivo_imagen: str,
    patente_imagen: str,
    ratio: float
  ) -> None:
    self.archivo_imagen: str = archivo_imagen
    self.patente_imagen: str = patente_imagen
    self.ratio: float = ratio


class Multa:
  """
  Representa una infracción por exceso de velocidad.
  """

  def __init__(
    self,
    multa_id: int | str,
    fecha: str,
    hora: str,
    velocidad_registrada: float,
    velocidad_maxima: float,
    estado_multa: str,
    exceso_velocidad_real: float,
    exceso_velocidad: float,
    vehiculo: Vehiculo,
    radar: Radar,
    evidencia: Evidencia | None = None
  ) -> None:
    self.multa_id: int | str = multa_id
    self.fecha: str = fecha
    self.hora: str = hora
    self.velocidad_registrada: float = velocidad_registrada
    self.velocidad_maxima: float = velocidad_maxima
    self.estado_multa: str = estado_multa
    self.exceso_velocidad_real: float = exceso_velocidad_real
    self.exceso_velocidad: float = exceso_velocidad
    self.vehiculo: Vehiculo = vehiculo
    self.radar: Radar = radar
    self.evidencia: Evidencia | None = evidencia
````

````python
#@title Función procesar_fila_csv

def procesar_fila_csv(row: dict) -> Multa:
  """
  Recibe una fila del CSV y la mapea a una instancia de Multa.
  """

  vehiculo: Vehiculo = Vehiculo(
    patente=str(row["patente"])
  )

  radar: Radar = Radar(
    radar_id=str(row["radar_id"]),
    ubicacion=str(row["ubicacion"])
  )

  evidencia: Evidencia | None = None

  if row.get("imagen"):
    evidencia = Evidencia(
      archivo_imagen=str(row["imagen"]),
      patente_imagen=str(row["patente_imagen"]),
      ratio=float(row["ratio"])
    )

  multa: Multa = Multa(
    multa_id=row["multa_id"],
    fecha=str(row["fecha"]),
    hora=str(row["hora"]),
    velocidad_registrada=float(row["velocidad_registrada"]),
    velocidad_maxima=float(row["velocidad_maxima"]),
    estado_multa=str(row["estado_multa"]),
    exceso_velocidad_real=float(
      row.get("exceso_velocidad_real", 0)
    ),
    exceso_velocidad=float(
      row.get("exceso_velocidad", 0)
    ),
    vehiculo=vehiculo,
    radar=radar,
    evidencia=evidencia
  )

  return multa
````

# Modelo Relacional

## Vehiculo

| Campo | Tipo | Restricción |
|---|---|---|
| patente | String | PK |

Relaciones:
- Un Vehiculo puede tener muchas Multas.

---

## Radar

| Campo | Tipo | Restricción |
|---|---|---|
| radar_id | String | PK |
| ubicacion | String | |

Relaciones:
- Un Radar puede generar muchas Multas.

---

## Multa

| Campo | Tipo | Restricción |
|---|---|---|
| multa_id | String | PK |
| fecha | String | |
| hora | String | |
| velocidad_registrada | Float | |
| velocidad_maxima | Float | |
| estado_multa | String | |
| exceso_velocidad_real | Float | |
| exceso_velocidad | Float | |
| patente | String | FK -> Vehiculo.patente |
| radar_id | String | FK -> Radar.radar_id |

Relaciones:
- Muchas Multas pertenecen a un Vehiculo.
- Muchas Multas son generadas por un Radar.
- Una Multa puede tener una Evidencia.

---

## Evidencia

| Campo | Tipo | Restricción |
|---|---|---|
| evidencia_id | Integer | PK |
| archivo_imagen | String | |
| patente_imagen | String | |
| ratio | Float | |
| multa_id | String | FK -> Multa.multa_id, UNIQUE |

Relaciones:
- Una Evidencia pertenece a una Multa.

````python
#@title Modelo relacional con SQLAlchemy

Base = declarative_base()


class VehiculoModel(Base):
  """
  Modelo relacional para la entidad Vehiculo.
  """

  __tablename__ = "vehiculos"

  patente = Column(String, primary_key=True)

  multas = relationship(
    "MultaModel",
    back_populates="vehiculo"
  )

  def __repr__(self) -> str:
    return f"VehiculoModel(patente={self.patente})"


class RadarModel(Base):
  """
  Modelo relacional para la entidad Radar.
  """

  __tablename__ = "radares"

  radar_id = Column(String, primary_key=True)
  ubicacion = Column(String)

  multas = relationship(
    "MultaModel",
    back_populates="radar"
  )

  def __repr__(self) -> str:
    return (
      f"RadarModel(radar_id={self.radar_id}, "
      f"ubicacion={self.ubicacion})"
    )


class MultaModel(Base):
  """
  Modelo relacional para la entidad Multa.
  """

  __tablename__ = "multas"

  multa_id = Column(String, primary_key=True)
  fecha = Column(String)
  hora = Column(String)
  velocidad_registrada = Column(Float)
  velocidad_maxima = Column(Float)
  estado_multa = Column(String)
  exceso_velocidad_real = Column(Float)
  exceso_velocidad = Column(Float)

  patente = Column(
    String,
    ForeignKey("vehiculos.patente")
  )

  radar_id = Column(
    String,
    ForeignKey("radares.radar_id")
  )

  vehiculo = relationship(
    "VehiculoModel",
    back_populates="multas"
  )

  radar = relationship(
    "RadarModel",
    back_populates="multas"
  )

  evidencia = relationship(
    "EvidenciaModel",
    back_populates="multa",
    uselist=False
  )

  def __repr__(self) -> str:
    return (
      f"MultaModel(multa_id={self.multa_id}, "
      f"patente={self.patente}, "
      f"radar_id={self.radar_id})"
    )


class EvidenciaModel(Base):
  """
  Modelo relacional para la entidad Evidencia.
  """

  __tablename__ = "evidencias"

  evidencia_id = Column(Integer, primary_key=True)
  archivo_imagen = Column(String)
  patente_imagen = Column(String)
  ratio = Column(Float)

  multa_id = Column(
    String,
    ForeignKey("multas.multa_id"),
    unique=True
  )

  multa = relationship(
    "MultaModel",
    back_populates="evidencia"
  )

  def __repr__(self) -> str:
    return (
      f"EvidenciaModel(evidencia_id={self.evidencia_id}, "
      f"multa_id={self.multa_id}, "
      f"ratio={self.ratio})"
    )
````

En este ejercicio vamos a vincular las 2 bases de datos, la relacional y la vectorial.

Crear la Base De Datos Vectorial `patente_vectorial` para almacenar los datos utilizando el modelo base OpenClip. Ambas instalaciones se realizaron al inicio del notebook.

Para realizar el almacenamiento vamos a tomar el id del Vehículo y el vector de la imagen.

La idea es que al obtener una imagen podamos obtener por aproximación el id del vehiculo.

Poblar la Base De Datos.



````python
#@title Crear base de datos vectorial patente_vectorial

import chromadb
cliente_vectorial = chromadb.PersistentClient(
  path="patente_vectorial"
)

coleccion_patentes = cliente_vectorial.get_or_create_collection(
  name="patentes"
)

print("Base vectorial patente_vectorial creada correctamente.")
````

````python
#@title Cargar modelo OpenCLIP

import open_clip
import torch
from PIL import Image

device: str = "cuda" if torch.cuda.is_available() else "cpu"

modelo_clip, _, preprocesador_clip = open_clip.create_model_and_transforms(
  "ViT-B-32",
  pretrained="openai"
)

modelo_clip = modelo_clip.to(device)
modelo_clip.eval()

print(f"Modelo OpenCLIP cargado en {device}.")
````

````python
#@title Función para generar vector de imagen

def obtener_vector_imagen(ruta_imagen: str) -> list[float]:
  """
  Genera el vector de una imagen utilizando OpenCLIP.
  """

  imagen = Image.open(ruta_imagen).convert("RGB")
  tensor_imagen = preprocesador_clip(imagen).unsqueeze(0).to(device)

  with torch.no_grad():
    vector = modelo_clip.encode_image(tensor_imagen)
    vector = vector / vector.norm(dim=-1, keepdim=True)

  return vector.cpu().numpy()[0].tolist()
````

````python
#@title Poblar base vectorial con evidencias

registros_evidencia = (
  session.query(
    MultaModel.multa_id,
    MultaModel.patente,
    EvidenciaModel.archivo_imagen
  )
  .join(EvidenciaModel)
  .all()
)

cantidad_insertada: int = 0

for multa_id, patente, archivo_imagen in registros_evidencia:
  ruta_imagen: Path = Path(archivo_imagen)

  if not ruta_imagen.exists():
    continue

  vector_imagen: list[float] = obtener_vector_imagen(
    str(ruta_imagen)
  )

  coleccion_patentes.add(
    ids=[str(multa_id)],
    embeddings=[vector_imagen],
    metadatas=[{
      "patente": str(patente),
      "imagen": str(ruta_imagen)
    }]
  )

  cantidad_insertada += 1

print(f"Vectores insertados en patente_vectorial: {cantidad_insertada}")
````

Crear la función `buscar_patente_imagen` que recibe como parámetro una imagen y retorna todos los datos del vehículo.

````python
#@title Función buscar_patente_imagen

def buscar_patente_imagen(ruta_imagen: str) -> dict:
  """
  Busca una patente por similitud visual y retorna datos del vehículo.
  """

  ruta_original: Path = Path(ruta_imagen)

  if not ruta_original.exists():
    ruta_alternativa: Path = Path(
      "urban_flow/data/raw/imgs"
    ) / ruta_original.name

    if ruta_alternativa.exists():
      ruta_original = ruta_alternativa
    else:
      return {
        "error": "Imagen no encontrada",
        "ruta": ruta_imagen
      }

  vector_imagen: list[float] = obtener_vector_imagen(
    str(ruta_original)
  )

  resultado_busqueda: dict = coleccion_patentes.query(
    query_embeddings=[vector_imagen],
    n_results=1
  )

  if len(resultado_busqueda["ids"][0]) == 0:
    return {}

  metadata: dict = resultado_busqueda["metadatas"][0][0]
  patente_detectada: str = str(metadata["patente"])

  vehiculo: VehiculoModel | None = (
    session.query(VehiculoModel)
    .filter(VehiculoModel.patente == patente_detectada)
    .first()
  )

  if vehiculo is None:
    return {}

  multas: list[MultaModel] = (
    session.query(MultaModel)
    .filter(MultaModel.patente == vehiculo.patente)
    .all()
  )

  return {
    "patente": vehiculo.patente,
    "cantidad_multas": len(multas),
    "multas": [
      {
        "multa_id": multa.multa_id,
        "fecha": multa.fecha,
        "hora": multa.hora,
        "estado_multa": multa.estado_multa,
        "velocidad_registrada": multa.velocidad_registrada,
        "velocidad_maxima": multa.velocidad_maxima,
        "exceso_velocidad": multa.exceso_velocidad
      }
      for multa in multas
    ]
  }
````
