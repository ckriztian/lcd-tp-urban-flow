# Ejercicio 03 - Modelo Lógico

## Descripción

Se diseñó un modelo lógico orientado a objetos a partir de los datos presentes en el archivo `speeding_fines_image.csv`.

Las entidades identificadas son:

- Vehiculo
- Multa
- Radar
- Evidencia

---

## Entidad: Vehiculo

Representa al vehículo involucrado en una infracción.

### Atributos

| Atributo | Tipo |
|-----------|------|
| patente | str |

### Relaciones

- Un Vehiculo puede tener muchas Multas.

---

## Entidad: Multa

Representa una infracción por exceso de velocidad.

### Atributos

| Atributo | Tipo |
|-----------|------|
| multa_id | int |
| fecha | str |
| hora | str |
| velocidad_registrada | float |
| velocidad_maxima | int |
| estado_multa | str |
| exceso_velocidad_real | float |
| exceso_velocidad | float |

### Relaciones

- Una Multa pertenece a un Vehiculo.
- Una Multa puede estar asociada a un Radar.
- Una Multa puede tener una Evidencia.

---

## Entidad: Radar

Representa el dispositivo encargado de detectar la infracción.

### Atributos

| Atributo | Tipo |
|-----------|------|
| radar_id | str |
| ubicacion | str |

### Relaciones

- Un Radar puede generar muchas Multas.

---

## Entidad: Evidencia

Representa la evidencia fotográfica asociada a una multa.

### Atributos

| Atributo | Tipo |
|-----------|------|
| imagen | str |
| patente_imagen | str |
| ratio | float |

### Relaciones

- Una Evidencia pertenece a una Multa.

---

## Cardinalidades

| Relación | Cardinalidad |
|-----------|-------------|
| Vehiculo → Multa | 1 : N |
| Radar → Multa | 1 : N |
| Multa → Evidencia | 1 : 0..1 |

---

## Diagrama Simplificado

```text
Vehiculo (1)
    |
    |------ (N)
    |
  Multa
    |
    |------ (0..1)
    |
Evidencia

Radar (1)
    |
    |------ (N)
    |
  Multa
```

````python
#@title Modelo Lógico del Sistema de Multas de Tránsito.
class Vehiculo:
    """
    Representa un vehículo involucrado en una infracción.
    """

    def __init__(self, patente: str) -> None:
        self.patente: str = patente


class Radar:
    """
    Representa un radar que detecta infracciones.
    """

    def __init__(self, radar_id: str, ubicacion: str) -> None:
        self.radar_id: str = radar_id
        self.ubicacion: str = ubicacion


class Evidencia:
    """
    Representa la evidencia fotográfica asociada a una multa.
    """

    def __init__(
        self,
        imagen: str,
        patente_imagen: str,
        ratio: float
    ) -> None:
        self.imagen: str = imagen
        self.patente_imagen: str = patente_imagen
        self.ratio: float = ratio


class Multa:
    """
    Representa una infracción por exceso de velocidad.
    """

    def __init__(
        self,
        multa_id: int,
        fecha: str,
        hora: str,
        velocidad_registrada: float,
        velocidad_maxima: int,
        estado_multa: str,
        exceso_velocidad_real: float,
        exceso_velocidad: float,
        vehiculo: Vehiculo,
        radar: Radar,
        evidencia: Evidencia
    ) -> None:
        self.multa_id: int = multa_id
        self.fecha: str = fecha
        self.hora: str = hora
        self.velocidad_registrada: float = velocidad_registrada
        self.velocidad_maxima: int = velocidad_maxima
        self.estado_multa: str = estado_multa
        self.exceso_velocidad_real: float = exceso_velocidad_real
        self.exceso_velocidad: float = exceso_velocidad

        self.vehiculo: Vehiculo = vehiculo
        self.radar: Radar = radar
        self.evidencia: Evidencia = evidencia
````

# Ejercicio 04

> Puntos: 1

Completar la función `procesar_fila_csv` que recibe como parámetro un `dict` y debe cargar los datos de una fila del archivo `data/processed/speeding_fines_image.csv` y mapearla según las clases definidas en el punto anterior.

````python
#@title Mapeo de una fila CSV al modelo lógico de multas




def procesar_fila_csv(fila: dict[str, Any]) -> Multa:
    """
    Convierte una fila del archivo CSV en objetos del modelo lógico.
    """

    vehiculo = Vehiculo(
        patente=fila["patente"]
    )

    radar = Radar(
        radar_id=fila["radar_id"],
        ubicacion=fila["ubicacion"]
    )

    evidencia = Evidencia(
        imagen=fila["imagen"],
        patente_imagen=fila["patente_imagen"],
        ratio=fila["ratio"]
    )

    multa = Multa(
        multa_id=fila["multa_id"],
        fecha=fila["fecha"],
        hora=fila["hora"],
        velocidad_registrada=fila["velocidad_registrada"],
        velocidad_maxima=fila["velocidad_maxima"],
        estado_multa=fila["estado_multa"],
        exceso_velocidad_real=fila["exceso_velocidad_real"],
        exceso_velocidad=fila["exceso_velocidad"],
        vehiculo=vehiculo,
        radar=radar,
        evidencia=evidencia,
    )

    return multa
````

## Multa

| Campo | Tipo | Restricción |
|---------|---------|---------|
| multa_id | Integer | PK |
| fecha | String | |
| hora | String | |
| velocidad_registrada | Float | |
| velocidad_maxima | Integer | |
| estado_multa | String | |
| exceso_velocidad_real | Float | |
| exceso_velocidad | Float | |
| patente | String | FK -> Vehiculo.patente |
| radar_id | String | FK -> Radar.radar_id |

Relaciones:
- Muchas Multas pertenecen a un Vehiculo.
- Muchas Multas pueden ser generadas por un Radar.
- Una Multa puede tener una Evidencia.

````python
#@title Diseño del Modelo Relacional con SQLAlchemy ORM


Base = declarative_base()


class Vehiculo(Base):
    """
    Modelo ORM que representa un vehículo.
    """

    __tablename__ = "vehiculos"

    patente = Column(String, primary_key=True)

    multas = relationship(
        "Multa",
        back_populates="vehiculo"
    )

    def __repr__(self) -> str:
        return f"Vehiculo(patente='{self.patente}')"


class Radar(Base):
    """
    Modelo ORM que representa un radar.
    """

    __tablename__ = "radares"

    radar_id = Column(String, primary_key=True)
    ubicacion = Column(String)

    multas = relationship(
        "Multa",
        back_populates="radar"
    )

    def __repr__(self) -> str:
        return (
            f"Radar(radar_id='{self.radar_id}', "
            f"ubicacion='{self.ubicacion}')"
        )


class Multa(Base):
    """
    Modelo ORM que representa una multa por exceso de velocidad.
    """

    __tablename__ = "multas"

    multa_id = Column(Integer, primary_key=True)

    fecha = Column(String)
    hora = Column(String)

    velocidad_registrada = Column(Float)
    velocidad_maxima = Column(Integer)

    estado_multa = Column(String)

    exceso_velocidad_real = Column(Float)
    exceso_velocidad = Column(Float)

    patente = Column(
        String,
        ForeignKey("vehiculos.patente")
    )

    radar_id = Column(
        String,
        ForeignKey("radares.radar_id"),
        nullable=True
    )

    vehiculo = relationship(
        "Vehiculo",
        back_populates="multas"
    )

    radar = relationship(
        "Radar",
        back_populates="multas"
    )

    evidencia = relationship(
        "Evidencia",
        back_populates="multa",
        uselist=False
    )

    def __repr__(self) -> str:
        return (
            f"Multa(multa_id={self.multa_id}, "
            f"estado='{self.estado_multa}')"
        )


class Evidencia(Base):
    """
    Modelo ORM que representa una evidencia fotográfica.
    """

    __tablename__ = "evidencias"

    id = Column(Integer, primary_key=True)

    imagen = Column(String)
    patente_imagen = Column(String)
    ratio = Column(Float)

    multa_id = Column(
        Integer,
        ForeignKey("multas.multa_id"),
        unique=True
    )

    multa = relationship(
        "Multa",
        back_populates="evidencia"
    )

    def __repr__(self) -> str:
        return (
            f"Evidencia(imagen='{self.imagen}', "
            f"ratio={self.ratio})"
        )
````
